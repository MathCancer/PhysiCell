# PhysiCelldFBA Unit Test: Mass Conversion

```python
from pathlib import Path
from pctk import multicellds
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator
import numpy as np
import pandas as pd
```

## 1. Configure analysis inputs and constants
Set the output path and the minimal constants needed to reconstruct simulated biomass from cell dry volume.

```python
output_folder = Path("../output")
cell_density_g_per_ml = 1.04
substrate_zero_threshold = 1e-12

if not output_folder.exists():
    raise FileNotFoundError(f"Output folder not found: {output_folder.resolve()}")
```

## 2. Load and aggregate simulation outputs
Initialize the MultiCellDS reader, compute substrate time series, and reconstruct total biomass from the simulated dry cell volume so that biomass can vary continuously between division events.

```python
mcds_reader = multicellds.MultiCellDS(str(output_folder))
microenvironment_indices = {
    "oxygen": 4,
    "glucose": 5,
    "acetate": 6,
    "co2": 7,
}

mcds_reader
```

```python
substrate_rows = []

for t, a in mcds_reader.microenvironment_as_matrix_iterator():
    oxygen = np.where(
        a[microenvironment_indices["oxygen"]] < substrate_zero_threshold,
        0.0,
        a[microenvironment_indices["oxygen"]],
    ).mean()
    glucose = np.where(
        a[microenvironment_indices["glucose"]] < substrate_zero_threshold,
        0.0,
        a[microenvironment_indices["glucose"]],
    ).mean()
    acetate = np.where(
        a[microenvironment_indices["acetate"]] < substrate_zero_threshold,
        0.0,
        a[microenvironment_indices["acetate"]],
    ).mean()
    co2 = np.where(
        a[microenvironment_indices["co2"]] < substrate_zero_threshold,
        0.0,
        a[microenvironment_indices["co2"]],
    ).mean()

    substrate_rows.append((t / 60.0, glucose, acetate, oxygen, co2))

df_substrates = pd.DataFrame(
    substrate_rows,
    columns=["time", "glucose", "acetate", "oxygen", "co2"],
).set_index("time")

df_substrates.head()
```

```python
cell_rows = []

for t, a in mcds_reader.cells_as_matrix_iterator():
    n_cells = a.shape[0]
    total_volume_um3 = a[:, 4].sum()
    fluid_fraction = a[:, 11]
    dry_volume_um3 = np.sum(a[:, 4] * (1.0 - fluid_fraction))
    total_biomass_pg = dry_volume_um3 * cell_density_g_per_ml

    cell_rows.append((
        t / 60.0,
        n_cells,
        total_volume_um3,
        dry_volume_um3,
        total_biomass_pg,
    ))

df_cells = pd.DataFrame(
    cell_rows,
    columns=["time", "n_cells", "total_volume_um3", "dry_volume_um3", "total_biomass_pg"],
).set_index("time")

df_cells.head()
```

```python
df_analysis = df_substrates.join(df_cells, how="inner")

summary_df = pd.DataFrame(
    {
        "initial": df_analysis.iloc[0],
        "final": df_analysis.iloc[-1],
        "maximum": df_analysis.max(),
    }
).loc[["glucose", "acetate", "oxygen", "n_cells", "dry_volume_um3", "total_biomass_pg"]]

summary_df
```

## 3. Analytical solution

```python
domain_bounds_um = (-8.0, 8.0)
domain_volume_um3 = 4096.0
domain_volume_L = domain_volume_um3 * 1e-15
voxel_resolution_um = 1.6
initial_glucose_mM = 10.0
glucose_molecular_weight_g_per_mol = 180.0
dry_weight_yield_gdw_per_g_glc = 0.5197  # Ecoli_core aerobic biomass yield on glucose (matches FBA model)
dry_mass_fraction = 0.25  # = 1 - fluid_fraction, consistent with cell dry volume
reference_cell_volume_um3 = 1.3
fluid_fraction = 0.75
number_of_cells = 1

glucose_concentration_mol_per_L = initial_glucose_mM * 1e-3
print("Glucose concentration (mol/L)): ", glucose_concentration_mol_per_L)

glucose_moles_in_domain = glucose_concentration_mol_per_L * domain_volume_L
print("Glucose moles in the domain: ", glucose_moles_in_domain)

glucose_mass_g = glucose_moles_in_domain * glucose_molecular_weight_g_per_mol
print("Glucose mass in the domain (g): ", glucose_mass_g)


theoretical_dry_biomass_g = glucose_mass_g * dry_weight_yield_gdw_per_g_glc
print("Theoretical dry mass (g): ", theoretical_dry_biomass_g)

theoretical_wet_biomass_g = theoretical_dry_biomass_g / dry_mass_fraction
print("Theoretical wet weight (g): ", theoretical_wet_biomass_g)

theoretical_total_biovolume_mL = theoretical_wet_biomass_g / cell_density_g_per_ml
print("Theoretical total biovolume (mL) ", theoretical_total_biovolume_mL)

reference_cell_volume_mL = reference_cell_volume_um3 * 1e-12
print("Reference cell volume (mL): ", reference_cell_volume_mL)

theoretical_max_cells = theoretical_total_biovolume_mL / reference_cell_volume_mL
print("Theoretical max cells: ", theoretical_max_cells)

biomass_from_glucose_pg = theoretical_dry_biomass_g * 1e12
initial_biomass_pg = (
    reference_cell_volume_um3 * (1.0 - fluid_fraction) * number_of_cells * cell_density_g_per_ml
)
theoretical_max_biomass_pg = initial_biomass_pg + biomass_from_glucose_pg
print("Initial single-cell biomass (pg): ", initial_biomass_pg)
print("Biomass from glucose (pg): ", biomass_from_glucose_pg)
print("Theoretical total biomass (pg): ", theoretical_max_biomass_pg)


analytical_solution = pd.Series(
    {
        "domain_min_um": domain_bounds_um[0],
        "domain_max_um": domain_bounds_um[1],
        "domain_volume_um3": domain_volume_um3,
        "domain_volume_L": domain_volume_L,
        "voxel_resolution_um": voxel_resolution_um,
        "reference_cell_volume_um3": reference_cell_volume_um3,
        "cell_density_g_per_ml": cell_density_g_per_ml,
        "initial_glucose_mM": initial_glucose_mM,
        "glucose_concentration_mol_per_L": glucose_concentration_mol_per_L,
        "glucose_moles_in_domain": glucose_moles_in_domain,
        "glucose_mass_g": glucose_mass_g,
        "dry_weight_yield_gdw_per_g_glc": dry_weight_yield_gdw_per_g_glc,
        "theoretical_dry_biomass_g": theoretical_dry_biomass_g,
        "theoretical_dry_biomass_pg": theoretical_max_biomass_pg,
        "dry_mass_fraction": dry_mass_fraction,
        "theoretical_wet_biomass_g": theoretical_wet_biomass_g,
        "theoretical_total_biovolume_fL": theoretical_total_biovolume_mL * 1e12,
        "theoretical_max_cells": theoretical_max_cells,
    },
    name="analytical_solution",
)
```

```python
comparison_df = pd.DataFrame(
    {
        "simulated_initial": [
            df_analysis.iloc[0]["total_biomass_pg"],
            df_analysis.iloc[0]["n_cells"],
        ],
        "simulated_final": [
            df_analysis.iloc[-1]["total_biomass_pg"],
            df_analysis.iloc[-1]["n_cells"],
        ],
        "simulated_max": [
            df_analysis["total_biomass_pg"].max(),
            df_analysis["n_cells"].max(),
        ],
        "analytical_final": [
            analytical_solution["theoretical_dry_biomass_pg"],
            analytical_solution["theoretical_max_cells"],
        ],
    },
    index=["biomass_pg", "cell_count"],
)

comparison_df
```

## 4. Plot the simulated trajectories against the analytical endpoint prediction
Panel A shows the simulated glucose dynamics. Panel B shows the simulated biomass and cell count over time together with the final analytical biomass and cell-number predictions from the domain-scale calculation above.

```python
# Cell Systems–style figure: stacked panels; glucose & CO2 on separate axes (top),
# total biomass + cell count vs analytical endpoint (bottom).
COLOR_GLUCOSE = "#f1c232"
COLOR_CO2 = "#8e44ad"
COLOR_BIOMASS = "#000000"
COLOR_NCELLS = "#1f78b4"

SCALE = 5.0  # uniform scale-up of the exported figure (proportions unchanged)
FS_LABEL = 9 * SCALE
FS_TICK = 8 * SCALE
FS_LEGEND = 8.5 * SCALE
LW_DATA = 1.35 * SCALE
LW_LIMIT = 0.85 * SCALE
LW_SPINE = 0.75 * SCALE
Y_PAD_BIOMASS = 1.08
Y_PAD_NCELLS = 1.15

tmax_h = 4.0  # cut the time axis at 4 hours (set <= 0 for full range)

_CELL_SYSTEMS_RC = {
    "font.family": "sans-serif",
    "font.sans-serif": ["Arial", "Helvetica", "DejaVu Sans", "Liberation Sans"],
    "font.size": FS_TICK,
    "axes.linewidth": LW_SPINE,
    "axes.labelweight": "normal",
    "axes.titleweight": "normal",
    "xtick.major.width": LW_SPINE * 0.65,
    "ytick.major.width": LW_SPINE * 0.65,
    "xtick.major.size": 3.5 * SCALE,
    "ytick.major.size": 3.5 * SCALE,
    "legend.frameon": False,
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
}


def _style_primary_ax(ax):
    ax.set_facecolor("white")
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    for side in ("left", "bottom"):
        ax.spines[side].set_color("black")
        ax.spines[side].set_linewidth(LW_SPINE)
    ax.tick_params(axis="both", which="major", labelsize=FS_TICK, colors="black",
                   length=4, width=LW_SPINE * 0.7)
    ax.tick_params(axis="both", which="minor", length=0)
    ax.grid(False)


def _style_twin_ax(ax):
    ax.set_facecolor("white")
    ax.spines["top"].set_visible(False)
    ax.spines["left"].set_visible(False)
    ax.spines["bottom"].set_visible(False)
    ax.spines["right"].set_visible(True)
    ax.spines["right"].set_color("black")
    ax.spines["right"].set_linewidth(LW_SPINE)
    ax.tick_params(axis="y", which="major", labelsize=FS_TICK, colors="black",
                   length=4, width=LW_SPINE * 0.7)


def _add_panel_legend(ax, lines, ncol, y_offset):
    leg = ax.legend([ln for ln in lines], [ln.get_label() for ln in lines],
                    loc="upper center", bbox_to_anchor=(0.5, y_offset), ncol=ncol,
                    frameon=False, fontsize=FS_LEGEND, handlelength=1.8,
                    handletextpad=0.5, columnspacing=1.2, labelspacing=0.35,
                    borderaxespad=0.0)
    leg.set_zorder(5)
    for text in leg.get_texts():
        text.set_color("#111111")


df_plot = df_analysis[df_analysis.index <= tmax_h] if tmax_h and tmax_h > 0 else df_analysis

with plt.rc_context(_CELL_SYSTEMS_RC):
    fig_w_in = (85.0 / 25.4) * SCALE
    fig_h_in = 4.6 * SCALE
    fig, (ax_top, ax_bot) = plt.subplots(
        2, 1, figsize=(fig_w_in, fig_h_in), sharex=True, facecolor="white"
    )

    # (top) glucose (left axis) + CO2 (right axis)
    line_glc, = ax_top.plot(
        df_plot.index, df_plot["glucose"], color=COLOR_GLUCOSE,
        linewidth=LW_DATA, label="Glucose", zorder=3, clip_on=False,
    )
    ax_top.set_ylabel("Glucose (mM)", fontsize=FS_LABEL, color=COLOR_GLUCOSE, labelpad=3)
    ax_top.set_ylim(0, max(1.12 * float(df_plot["glucose"].max()), 0.1))
    _style_primary_ax(ax_top)

    ax_co2 = ax_top.twinx()
    line_co2, = ax_co2.plot(
        df_plot.index, df_plot["co2"], color=COLOR_CO2,
        linewidth=LW_DATA, label="CO₂", zorder=4, clip_on=False,
    )
    ax_co2.set_ylabel("CO₂ (mM)", fontsize=FS_LABEL, color=COLOR_CO2, labelpad=4)
    ax_co2.set_ylim(0, max(1.12 * float(df_plot["co2"].max()), 0.1))
    _style_twin_ax(ax_co2)

    # (bottom) total biomass (left axis) + cell count (right axis), with analytical endpoints
    line_biomass, = ax_bot.plot(
        df_plot.index, df_plot["total_biomass_pg"], color=COLOR_BIOMASS,
        linewidth=LW_DATA + 0.25, label="Simulated total biomass", zorder=3, clip_on=False,
    )
    line_analytical = ax_bot.axhline(
        theoretical_max_biomass_pg, color=COLOR_BIOMASS, linestyle=":",
        linewidth=LW_LIMIT, label="Analytical total biomass", zorder=2,
    )
    ax_bot.set_xlabel("Time (h)", fontsize=FS_LABEL, color="black", labelpad=4)
    ax_bot.set_ylabel("Total biomass (pg DW)", fontsize=FS_LABEL, color="black", labelpad=3)
    ax_bot.set_ylim(
        0,
        Y_PAD_BIOMASS * max(
            float(df_plot["total_biomass_pg"].max()), float(theoretical_max_biomass_pg)
        ),
    )
    _style_primary_ax(ax_bot)

    ax_ncells = ax_bot.twinx()
    line_ncells, = ax_ncells.step(
        df_plot.index, df_plot["n_cells"], color=COLOR_NCELLS, where="post",
        linewidth=LW_DATA, linestyle="--", label="Simulated cell count", zorder=4, clip_on=False,
    )
    ax_ncells.set_ylabel("Cell count", fontsize=FS_LABEL, color=COLOR_NCELLS, labelpad=4)
    ax_ncells.set_ylim(
        0,
        Y_PAD_NCELLS * max(
            float(df_plot["n_cells"].max()), float(analytical_solution["theoretical_max_cells"])
        ),
    )
    ax_ncells.yaxis.set_major_locator(MaxNLocator(integer=True))
    _style_twin_ax(ax_ncells)

    _add_panel_legend(
        ax_bot, [line_biomass, line_analytical, line_ncells], ncol=1, y_offset=-0.40
    )

    x_max = float(tmax_h) if (tmax_h and tmax_h > 0) else float(df_plot.index.max())
    ax_bot.set_xlim(float(df_plot.index.min()), x_max)

    fig.subplots_adjust(left=0.16, right=0.84, bottom=0.24, top=0.97, hspace=0.35)
    fig.align_ylabels([ax_top, ax_bot])

plt.show()
```
